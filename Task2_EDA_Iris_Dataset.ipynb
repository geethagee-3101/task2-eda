```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame
df["species"] = df["target"].map(dict(enumerate(iris.target_names)))
df = df.drop(columns=["target"])

df.head()
```

```python

print("Dataset shape:", df.shape)
print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())
```

```python

df.describe()
```

```python

sns.histplot(data=df, x="sepal length (cm)", kde=True)
plt.title("Distribution of Sepal Length")
plt.show()
```

```python

sns.boxplot(data=df, x="petal length (cm)")
plt.title("Petal Length Outliers")
plt.show()
```

```python

sns.scatterplot(data=df, x="petal length (cm)",
                y="petal width (cm)", hue="species")
plt.title("Petal Length vs Petal Width")
plt.show()
```

```python

sns.heatmap(df.select_dtypes("number").corr(),
            annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python

sns.countplot(data=df, x="species")
plt.title("Number of Flowers by Species")
plt.show()
```

Top 3 Insights

1. Petal length and petal width have a strong positive correlation.
2. Setosa flowers are clearly separated from the other species based on petal measurements.
3. Each species has 50 samples, so the dataset is balanced.
